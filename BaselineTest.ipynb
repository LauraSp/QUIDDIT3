```python
import QUtility as QU
import matplotlib.pyplot as plt

std_folder = r'test/subset/linescan/standard/'
std_files = [f for f in os.listdir(std_folder) if f.endswith('.csv')]

exp_folder = r'test/subset/linescan/experimental/'
exp_files = [f for f in os.listdir(exp_folder) if f.endswith('.csv')]

for file in std_files:
    std_spec = QU.QUtility.read_spec(os.path.join(std_folder, file))
    exp_spec = QU.QUtility.read_spec(os.path.join(exp_folder, file))

    plt.figure()
    plt.plot(std_spec[:, 0], std_spec[:, 1], label='standard')
    plt.plot(exp_spec[:, 0], exp_spec[:, 1], label='experimental')
    plt.plot(std_spec[:, 0], std_spec[:, 1] - exp_spec[:, 1], label='difference')
    plt.legend()
```

## Testing map

```python
import os
import sys

# Now import your module
from QBaseline import remove_baseline

spectra_folder = r'test/map/'

# Get list of spectra ending in 0.csv or 5.csv
spectra_files = [f for f in os.listdir(spectra_folder) if f.endswith('0.csv') or f.endswith('5.csv')]
```

### 1) Correct baseline

```python
for spec in spectra_files:
    spec_path = os.path.join(spectra_folder, spec)
    remove_baseline(spec_path, spectra_folder, bl_type='experimental')
```

### Plot corrected spectra together

```python
import QUtility as QU
import matplotlib.pyplot as plt

std_folder = r'test/map/standard/'
std_files = [f for f in os.listdir(std_folder) if f.endswith('.csv')]

exp_folder = r'test/map/experimental/'
exp_files = [f for f in os.listdir(exp_folder) if f.endswith('.csv')]

for file in exp_files:
    exp_spec = QU.QUtility.read_spec(os.path.join(exp_folder, file))
    # this requires some pattern matching. The experimental filenames may have a
    # prefic noise_ or failed_ rather than c
    try:
        std_spec = QU.QUtility.read_spec(os.path.join(std_folder, file))
    except FileNotFoundError:
        # remove leading noise_ or failed_ and try again
        std_file = file
        if std_file.startswith('noise_'):
            std_file = 'c' + std_file[len('noise_'):]
        if std_file.startswith('failed_'):
            std_file = 'c' + std_file[len('failed_'):]
        try:
            std_spec = QU.QUtility.read_spec(os.path.join(std_folder, std_file))
        except FileNotFoundError:
            print(f"File not found: {file}")
            continue

    plt.figure()
    #plt.plot(std_spec[:, 0], std_spec[:, 1], label='standard')
    #plt.plot(exp_spec[:, 0], exp_spec[:, 1], label='experimental')
    plt.plot(std_spec[:, 0], std_spec[:, 1] - exp_spec[:, 1], label='difference')
    plt.legend()
    plt.gca().invert_xaxis()
```
